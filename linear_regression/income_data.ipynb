```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error,mean_squared_error,r2_score
import matplotlib.pyplot as plt 

```

```python
df=pd.read_csv(r"C:\Users\User\Desktop\me journey\aml_lab_preparation\data\income_data.csv")
df.head()
```

```python
x=df.iloc[:,[1]].values
y=df.iloc[:,-1].values
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.25,random_state=0)
```

```python
model=LinearRegression()
model.fit(x_train,y_train)
```

```python
y_pred=model.predict(x_test)
```

```python
print(model.intercept_)
print(model.coef_[0])
```

```python
import numpy as np
mae=mean_absolute_error(y_test,y_pred)
mse=mean_squared_error(y_test,y_pred)
rmse=np.sqrt(mse)
r_square=r2_score(y_test,y_pred)
print(mae)
print(mse)
print(rmse)
print(r_square)
```

```python
plt.scatter(x_train,y_train,color="blue")
plt.scatter(x_test,y_test,color="green")
plt.plot(x_test,y_pred,color="red")
plt.xlabel("income")
plt.ylabel("happyness")
plt.grid()
plt.show()
```

```python
resuidial=y_test-y_pred
plt.scatter(y_pred,resuidial,color="red")
plt.axhline(0,color="blue")
plt.xlabel("happeniss")
plt.ylabel("resuidial")
plt.legend()
plt.show()
```

```python
plt.scatter(y_test, y_pred, color="blue")

minimum = min(y_test.min(), y_pred.min())
maximum = max(y_test.max(), y_pred.max())

plt.plot(
    [minimum, maximum],
    [minimum, maximum],
    color="red",
    linestyle="--"
)

plt.xlabel("actual happiness")
plt.ylabel("predicted happiness")
plt.title("actual vs predicted")
plt.show()
```
