```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn import metrics
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.decomposition import PCA

df=pd.read_csv(r"C:\Users\User\Desktop\me journey\aml_lab_preparation\data\Wine.csv")
df.head()


x=df.iloc[:,0:13].values
y=df.iloc[:,-1].values

x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.25,random_state=0)

sc = StandardScaler()

x_train = sc.fit_transform(x_train)
x_test = sc.transform(x_test)

pca=PCA(n_components=2)

x_train=pca.fit_transform(x_train)
x_test=pca.transform(x_test)

explained_varaince=pca.explained_variance_
print(explained_varaince)

model=LogisticRegression()
model.fit(x_train,y_train)
y_pred=model.predict(x_test)

cm = metrics.confusion_matrix(y_test, y_pred)
print(cm)

prediction = metrics.accuracy_score(y_test, y_pred)
print("Accuracy:", prediction * 100, "%")

print(metrics.classification_report(y_test, y_pred))
```

```python
l2_model=LogisticRegression(
    penalty='l2',
    C=1.0,
    random_state=0
)

l2_model.fit(x_train,y_train)
l2_pred=l2_model.predict(x_test)

cm_l2 = metrics.confusion_matrix(y_test, l2_pred)
print(cm_l2)

print("L2 Accuracy:", metrics.accuracy_score(y_test, l2_pred) * 100, "%")

print(metrics.classification_report(y_test, l2_pred))


l1_model = LogisticRegression(
    penalty="l1",
    solver="saga",
    C=1.0,
    random_state=0
)

l1_model.fit(x_train, y_train)

l1_pred = l1_model.predict(x_test)

cm_l1 = metrics.confusion_matrix(y_test, l1_pred)
print(cm_l1)

print("L1 Accuracy:", metrics.accuracy_score(y_test, l1_pred) * 100, "%")

print(metrics.classification_report(y_test, l1_pred))


```

```python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression(
    solver="lbfgs",
    max_iter=1000
)

model.fit(x_train, y_train)

y_pred=model.predict(x_test)

cm = metrics.confusion_matrix(y_test, y_pred)
print(cm)

prediction = metrics.accuracy_score(y_test, y_pred)
print("Accuracy:", prediction * 100, "%")

print(metrics.classification_report(y_test, y_pred))
```
