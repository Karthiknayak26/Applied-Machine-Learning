```python
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error,mean_squared_error,r2_score
import numpy as np
import matplotlib.pyplot as plt
```

```python
data={
    "study_hours":[2, 3, 4, 5, 6, 7, 8, 9],
    "attendance":[60, 65, 70, 75, 80, 85, 90, 95],
    "assignment_score":[50, 55, 58, 65, 70, 74, 80, 88],
    "final_marks":[42, 48, 54, 61, 67, 73, 81, 89]
}
df=pd.DataFrame(data)
df.head()
```

```python
x=df.iloc[:,[0,1,2]].values
y=df.iloc[:,-1].values

x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.25,random_state=42)
```

```python
import numpy as np
model=LinearRegression()
model.fit(x_train,y_train)
y_pred=model.predict(x_test)
print(model.intercept_)
print(model.coef_)
coefficent_table=pd.DataFrame({
    
    "coefiicient":model.coef_
})
print(coefficent_table)

mae=mean_absolute_error(y_test,y_pred)
mse=mean_squared_error(y_test,y_pred)
rmse=np.sqrt(mse)
r_square=r2_score(y_test,y_pred)
print(mae)
print(mse)
print(rmse)
print(r_square)
plt.scatter(y_test,y_pred,color="blue")
minimum=min(y_test.min(),y_pred.min())
maximum=max(y_test.max(),y_pred.max())
plt.plot(
    [minimum,maximum],
    [minimum,maximum],
    color="red"
)
plt.xlabel("income")
plt.ylabel("happyness")
plt.grid()
plt.show()
resuidial=y_test-y_pred
plt.scatter(y_pred,resuidial,color="red")
plt.axhline(0,color="blue")
plt.xlabel("happeniss")
plt.ylabel("resuidial")
plt.legend()
plt.show()
```
